```python
# Load the data from the filee
from pypdf import PdfReader
def pdfParser(filePath):
  all_text=[]
  reader = PdfReader(filePath)
  for i in reader.pages:
    page = i
    all_text.append(page.extract_text())
  all_text="\n".join(all_text)
  return {
      "text":all_text,
      "numPages":reader.get_num_pages()
  }

pdfdata=pdfParser("pdf.pdf")
```

```python
# download langchain to use recursivetextsplitter
!pip install langchain
!pip install -U langchain-text-splitters
```

```python
# organize the data , chunk it into parent and child,embend it
# this would be changed based on the embedding model
from langchain_text_splitters import RecursiveCharacterTextSplitter
import uuid
def chunkText(text,source):
  splitted_text=RecursiveCharacterTextSplitter(chunk_size=800, chunk_overlap=150)
  chunks=splitted_text.split_text(text)
  doc_id=uuid.uuid4()
  data=[
      {
          "chunk_id":str(doc_id),
          "doc_id":str(doc_id),
          "source":source,
          "content":chunk
      }
      for index,chunk in enumerate(chunks)
  ]
  return data

chunkTexts=chunkText(pdfdata['text'],'pdf.pdf')
"""
OUTPUT EXAMPLE:
[
  {
  'chunk_id': '3cfae791-280c-4829-8d17-a52d5f1522bd',
  'doc_id': '3cfae791-280c-4829-8d17-a52d5f1522bd',
  'source': 'pdf.pdf',
  'content': 'UNDERSTANDING MENTAL HEALTH\nWHAT IS MENTAL HEALTH?\'
  },.....
]
 """
print(len(chunkTexts))
```

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer('all-MiniLM-L6-v2')
```

```python
# vectorize the data
BATCH_SIZE=10
def embeddingText(text):
 return model.encode(text,batch_size=BATCH_SIZE,
        show_progress_bar=True)

# def embedBatch(texts):
#   vectors=[]
#   for textIndex in range(0,len(texts),BATCH_SIZE):
#     vectors.append(embeddingText(texts[textIndex]))
#   return vectors.extend()

def embendChunks(chunks):
  # take the text only from the chunks
  texts=[chunk["content"] for chunk in chunks]
  vectors=embeddingText(texts)
  result=[
      {
          **chunk,
          "vector":vectors[index].tolist()
      }

      for index,chunk in enumerate(chunks)
  ]
  return result


embendedChunks=embendChunks(chunkTexts)

# """
# OUTPUT EXAMPLE
# [
#   {

# 'chunk_id': '5f1fd414-9ac2-4e4b-9414-4dba4955f33d',
#   'doc_id': '5f1fd414-9ac2-4e4b-9414-4dba4955f33d',
#   'source': 'pdf.pdf',
#   'content': 'Reword any sentence that uses \npsychiatric or medical terminology \nincorrectly or out of conte',
#   'vector': [0.04775984212756157,
#    0.08367247879505157,
#    -0.028460843488574028,]

#    },etc....
#   ]
# """
```

```python
# ------Setup Weviate vector db-----

import weaviate
from weaviate.classes.config import Configure
from weaviate.classes.init import Auth
import weaviate.classes as wvc
from weaviate.classes.config import Configure, DataType, Property
client = weaviate.connect_to_weaviate_cloud(
    cluster_url=WEAVIATE_URL,
    auth_credentials=Auth.api_key(WEVIATE_API_KEY),
)
# create predefined properties
properties=[
   Property(name="chunk_id", data_type=DataType.TEXT),
    Property(name="doc_id", data_type=DataType.TEXT),
    Property(name="source", data_type=DataType.TEXT),
    Property(name="content", data_type=DataType.TEXT),
]


# create cluster in case it doesn' exist
if client.is_ready():
  if not client.collections.exists("llmlearning")==False:
    client.collections.create(
    name="llmlearning",
     vector_config=Configure.Vectorizer.none(),
    properties=properties
    )
    print("Collection is ready!")
else:
  print("Cluster is not ready!")
collection = client.collections.get("llmlearning");
```

```python
# from weaviate.classes import wvc
# Store the embeddings in db weviate in batch as well
WEVIATE_BATCH_SIZE=10
all_results=[]
def weviate_embend_chunks(chunks):
  for index in range(0,len(chunks),WEVIATE_BATCH_SIZE):
    # take batch_size
    batch=chunks[index:index+WEVIATE_BATCH_SIZE]
    # EMBEND THOS CHUNKS IN WEVAITE
    batch_to_insert=[
          wvc.data.DataObject(
                    properties=
           {
               "chunk_id":chunk["chunk_id"],
               "doc_id":chunk["doc_id"],
               "source":chunk["source"],
               "content":chunk["content"],
           },
           vector=chunk["vector"]

                              )

    for chunk in batch
    ]

    result=collection.data.insert_many(
        batch_to_insert,
        # show_progress_bar=True
    )
    if result.has_errors:
            print(f"Errors occurred in batch {index}: {result.errors}")
    all_results.append(result)
  return all_results



weviate_embend_chunks(embendedChunks)


```

```python
# Create similiarty search function
from weaviate.classes.query import MetadataQuery
def similarity_search(user_query_vector,top_k):
  result=collection.query.near_vector(user_query_vector,
                               limit=top_k,
                               return_metadata=MetadataQuery(distance=True),
                               return_properties=["chunk_id","doc_id","source","content"]
                               )
  return result
```

```python
# user query and retrieval data top_k
user_query="what is mental health"
top_k=4
score=0.5
user_query_vector=embeddingText(user_query)
result=similarity_search(user_query_vector,top_k)
for o in result.objects:
    if o.metadata.distance<score:
      print(o.properties)
```
